```python
reader=PdfReader('Review Paper.pdf')
len(reader.pages)
text=''
for page in reader.pages:
    text=text+page.extract_text()
```

```python
text
len(text)  #character count
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
chunks=splitter.split_text(text)
```

```python
len(chunks)
len(chunks[1])
```

```python
import chromadb
```

```python
client=chromadb.Client()
collection=client.get_or_create_collection('Android_pdf')
```

```python
collection.add(documents=chunks,ids=['c' + str(i) for i in range(len(chunks))])
```

```python
question='what is the review paper about?'
result=collection.query(query_texts=[question],n_results=2)
```

```python
context=result['documents'][0][1]
```

```python
llm=ChatGroq(model='openai/gpt-oss-120b')
prompt='''I will provide you question and the context which may have the answer of the given question .You have to generate answer in 2 lines with respect to contextYou have to generate answer with respect to given context.If you dont find answer in given context,just reply - Answer not found in given context.question:{}
context:{}'''.format(question,context)
result=llm.invoke(prompt)
print(result.content)

```
